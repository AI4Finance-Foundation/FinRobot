# Tutorial 1: Equity Research Report

Generate a comprehensive equity research report for any public company using FinRobot.

**What you'll learn:**
- How to initialize FinRobot with a model
- How to run the equity research pipeline
- How to interpret PipelineResult output

**Requirements:**
- `pip install finrobot`
- An LLM API key (set `FINAGENT_DEEPSEEK_API_KEY` or configure another provider)

````python
from finrobot import FinRobot

# Initialize with your preferred model
# Options: "deepseek:deepseek-chat", "openai:gpt-4o", "anthropic:claude-sonnet-4-6"
agent = FinRobot(model="deepseek:deepseek-chat")
````

## Run the Pipeline

The equity research pipeline runs 5 enforced steps:
1. **Data Collection** — fetches financials, price history, SEC filings
2. **Peer Analysis** — identifies and benchmarks comparable companies
3. **Financial Modeling** — runs DCF valuation with sensitivity analysis
4. **Thesis Generation** — synthesizes buy/hold/sell recommendation
5. **Report Writing** — produces the final research report

````python
# Run equity research on Apple (takes 30-120 seconds)
result = await agent.aresearch("AAPL")

# Print the summary
print(result.format_summary())
````

## Inspect Individual Steps

Each pipeline step's output is available in `result.steps`.

````python
# See which steps ran
print("Steps completed:", list(result.steps.keys()))

# Access structured data (typed Pydantic models)
for key, model in result.structured_data.items():
    print(f"  {key}: {type(model).__name__}")
````

## View the HTML Report

To view the full HTML report with charts, use the server:
```bash
finrobot serve
```
Then trigger analysis via the API and visit `/api/report/html?ticker=AAPL`.

````python
# Clean up
await agent.close()
````
