# Tutorial 4: RAG Q&A over SEC Filings

Ask questions about a company's 10-K filing with source citations.

**What you'll learn:**
- How to use `aask()` for RAG-based Q&A
- How BM25 retrieval works under the hood
- How to use standalone financial analysis

```python
from finrobot import FinRobot

agent = FinRobot(model="deepseek:deepseek-chat")
```

## Ask About Risk Factors

The `aask()` method:
1. Fetches the company's latest 10-K from SEC EDGAR
2. Extracts key sections (Business, Risk Factors, MD&A, Financial Statements)
3. Chunks the text and builds a BM25 index
4. Retrieves the most relevant passages for your question
5. Sends passages + question to the LLM
6. Returns an answer with source citations

```python
answer = await agent.aask("AAPL", "What are the main risk factors mentioned in the 10-K?")
print(answer)
```

## Ask About Strategy

```python
answer = await agent.aask("AAPL", "How did management discuss AI strategy in the latest filing?")
print(answer)
```

## Ask About Financials

```python
answer = await agent.aask("AAPL", "What was the revenue breakdown by segment?")
print(answer)
```

## Standalone Financial Analysis

Use `aanalyze()` for focused analysis without the full pipeline.

```python
# Income statement analysis
analysis = await agent.aanalyze("AAPL", "income")
print(analysis)
```

```python
# Risk assessment
risk = await agent.aanalyze("AAPL", "risk")
print(risk)
```
